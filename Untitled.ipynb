```python
from pathlib import Path
from pypdf import PdfReader

# Trouver la racine du projet, peu importe d'où on lance le notebook
ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

PDF_PATH = ROOT / "data" / "raw" / "ai_act_fr.pdf"
print("Je lis :", PDF_PATH)

reader = PdfReader(PDF_PATH)
print("Nombre de pages :", len(reader.pages))
print(reader.pages[20].extract_text()[:2000])
```

```python
import re

# Tout le texte du PDF en une seule chaîne
texte = "\n".join(page.extract_text() for page in reader.pages)
print("Nombre de caractères :", len(texte))

# 1. Où commence l'article 1 ?
idx = texte.find("Article premier")
print(texte[idx - 300 : idx + 1200])
print("=" * 80)

# 2. Combien de titres d'articles ? (on s'attend à 113)
articles = re.findall(r"^Article (premier|\d+)\s*$", texte, re.MULTILINE)
print("Articles trouvés :", len(articles))
print("Premiers :", articles[:5], "| Derniers :", articles[-5:])

# 3. Combien de considérants ? (on s'attend à 180)
considerants = re.findall(r"^\((\d+)\)", texte, re.MULTILINE)
print("Considérants trouvés :", len(considerants))
```

```python
vus = set()
for m in re.finditer(r"^\((\d+)\)(.*)$", texte, re.MULTILINE):
    n = int(m.group(1))
    if n in vus:
        print(f"({n}) déjà vu →", m.group(2)[:90])
    vus.add(n)
```

```python
# 1. Les considérants sont uniquement dans le préambule
preambule = texte[: texte.find("ONT ADOPTÉ LE PRÉSENT RÈGLEMENT")]

# 2. Une note de bas de page commence par une référence à un autre texte
NOTE = re.compile(
    r"^\(\d+\)\s+(JO [CL] |Règlement|Directive|Décision|Recommandation|Résolution|Conseil européen)"
)

considerants = [
    int(m.group(1))
    for m in re.finditer(r"^\((\d+)\)(.*)$", preambule, re.MULTILINE)
    if not NOTE.match(m.group(0))
]

print("Considérants :", len(considerants))
print("Numérotation parfaite (1 à 180) ?", considerants == list(range(1, 181)))
```

```python
from collections import Counter

compte = Counter(considerants)
doublons = [n for n, k in compte.items() if k > 1]
print("Numéros en double :", doublons)

# Afficher toutes les lignes qui portent ce numéro
for m in re.finditer(r"^\((\d+)\)(.*)$", preambule, re.MULTILINE):
    if int(m.group(1)) in doublons and not NOTE.match(m.group(0)):
        print(m.group(0)[:150])
```

```python
NOTE = re.compile(
    r"^\(\d+\)\s+(JO [CL] |Règlement|Directive|Décision|Recommandation|Résolution|Conseil européen|Position)"
)

considerants = [
    int(m.group(1))
    for m in re.finditer(r"^\((\d+)\)(.*)$", preambule, re.MULTILINE)
    if not NOTE.match(m.group(0))
]

print("Considérants :", len(considerants))
print("Numérotation parfaite (1 à 180) ?", considerants == list(range(1, 181)))
```
